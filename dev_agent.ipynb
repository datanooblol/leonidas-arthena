```python
class QueryDetectionAgent:
    def detect(self, message: str) -> dict:
        # Rule-based detection
        sql_indicators = [
        'แสดง', 'ดู', 'หา', 'ค้นหา', 'เอา', 'ขอ', 'ต้องการ', 'show', 'display', 'get', 'find', 'retrieve', 'list', 'give',
        
        # คำถามเชิงปริมาณ
        'เท่าไหร่', 'กี่', 'จำนวน', 'รวม', 'ทั้งหมด', 'ยอด', 'how many', 'how much', 'total', 'sum', 'count', 'amount',
        
        # คำถามเชิงวิเคราะห์
        'วิเคราะห์', 'เปรียบเทียบ', 'เฉลี่ย', 'สูงสุด', 'ต่ำสุด', 'แนวโน้ม', 'analyze', 'compare', 'average', 'avg', 'max', 'min', 'trend', 'insight',
        
        # # รูปแบบคำถาม
        # 'อะไร', 'ยังไง', 'ทำไม', 'ที่ไหน', 'เมื่อไหร่', 'what', 'how', 'why', 'where', 'when', 'who',
        ]
        is_query = any(indicator in message.lower() for indicator in sql_indicators)
        
        return {
            'is_query': is_query,
        }
```

```python
# message = 'อยากรู้ยอดขายรวมของเดือนมกราคม'
message = conversation_history[-2]['content']
QueryDetectionAgent().detect(message=message)
```
